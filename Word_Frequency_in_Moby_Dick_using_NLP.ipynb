# **Step 2 : Import Libraries**

```python
import requests
from bs4 import BeautifulSoup
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from collections import Counter
import pandas as pd
import matplotlib.pyplot as plt
from wordcloud import WordCloud
import string
```

# **Step 3 : Download NLTK Resources**

```python
nltk.download('punkt')
nltk.download('stopwords')
```

# **Step 4 : Scrape the Moby Dick Text**

We will scrape the novel from Project Gutenberg.

```python
url = "https://www.gutenberg.org/files/2701/2701-0.txt"

response = requests.get(url)
text = response.text

print(text[:1000])
```

# **Step 5 : Text Preprocessing**

**Convert Text to Lowercase**

```python
text = text.lower()
```

**Tokenize Words**

```python
nltk.download('punkt_tab')
tokens = word_tokenize(text)
```

**Remove Punctuation**

```python
words = [word for word in tokens if word.isalpha()]
```

**Remove Stopwords**

```python
stop_words = set(stopwords.words('english'))

filtered_words = [word for word in words if word not in stop_words]
```

# **Step 6 : Calculate Word Frequencies**

```python
word_freq = Counter(filtered_words)

print(word_freq.most_common(20))
```

# **Step 7 : Create Frequency DataFrame**

```python
df = pd.DataFrame(word_freq.most_common(20),
                  columns=['Word', 'Frequency'])

print(df)
```

# **Step 8 : Visualize Top 20 Words**

```python
plt.figure(figsize=(12,6))

plt.bar(df['Word'], df['Frequency'])

plt.title('Top 20 Most Frequent Words in Moby Dick')
plt.xlabel('Words')
plt.ylabel('Frequency')

plt.xticks(rotation=45)

plt.show()
```

# **Step 9 : Generate Word Cloud**

```python
wordcloud = WordCloud(
    width=1000,
    height=500,
    background_color='white'
).generate_from_frequencies(word_freq)

plt.figure(figsize=(15,8))

plt.imshow(wordcloud, interpolation='bilinear')

plt.axis('off')

plt.title("Word Cloud of Moby Dick")

plt.show()
```
